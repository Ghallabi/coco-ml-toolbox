```python
import os
os.sys.path.append("..")
from cocomltools.coco_metrics import COCOMetrics
from pathlib import Path
```

```python
coco_path_gt = Path("...")
coco_path_preds = Path("...")
coco_metrics = COCOMetrics(ground_truth_coco_file=coco_path_gt, predictions_coco_file=coco_path_preds)
stats, per_class_metrics = coco_metrics.evaluate()
```

```python
import numpy as np
ious = np.arange(0.5, 0.95, 0.05)
```

```python
precisions = []
for iou in ious:
    overall_stats, per_class_stats = coco_metrics.evaluate(iou_threshold=iou)
    precisions.append(overall_stats['Precision'])
print(f"Mean average precision@0.5-0.95: {round(sum(precisions) / len(precisions), 3)}")
```
